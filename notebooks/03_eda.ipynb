# 03 — Exploratory Data Analysis
Use charts to understand employee, attendance, work-hour and salary patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
payroll = pd.read_csv('../data/processed/payroll_processed.csv')
attendance = pd.read_csv('../data/processed/cleaned_attendance.csv', parse_dates=['Date'])
```

```python
fig, ax = plt.subplots(figsize=(10,5))
payroll.groupby('Department')['Net_Salary'].mean().sort_values().plot(kind='barh', ax=ax)
ax.set_title('Average Net Salary by Department')
ax.set_xlabel('Average Net Salary')
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,5))
sns.histplot(payroll['Attendance_Percentage'], bins=15, kde=True, ax=ax)
ax.set_title('Attendance Percentage Distribution')
ax.set_xlabel('Attendance Percentage')
plt.tight_layout()
plt.show()
```
