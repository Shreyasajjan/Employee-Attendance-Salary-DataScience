# 05 — Statistical Analysis
Correlation and group-level analysis. Statistical significance must be interpreted carefully and does not imply causation.

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr
p = pd.read_csv('../data/processed/payroll_processed.csv')
```

```python
print('Pearson attendance vs net salary:', pearsonr(p['Attendance_Percentage'], p['Net_Salary']))
print('Spearman attendance vs net salary:', spearmanr(p['Attendance_Percentage'], p['Net_Salary']))
```

```python
p.groupby('Department')[['Monthly_Salary','Net_Salary','Attendance_Percentage']].agg(['mean','median','std']).round(2)
```
